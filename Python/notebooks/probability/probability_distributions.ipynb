```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path

# Project root toevoegen aan Python path
PROJECT_ROOT = Path.cwd().resolve().parent
print(PROJECT_ROOT)
sys.path.insert(0, str(PROJECT_ROOT))

from statistics_utils.calculations.distributions import (
    get_distribution,
    CustomContinuousDistribution,
    probability,
    boundary_values,
    compute_expected_value,
    compute_variance,
    compute_standard_deviation,
)
```

## Willekeurige discrete verdelingen

```python
blood_units = {
    1: 0.2,
    2: 0.32,
    3: 0.288,
    4: 0.1536,
    5: 0.0384,
}
```

### 1. Kansen uitrekenen

```python
print(f"P(X = 2) = {probability(blood_units, lower=2, upper=2):.4f}")
print(f"P(X <= 2) = {probability(blood_units, upper=2):.4f}")
print(f"P(X >= 2) = {probability(blood_units, lower=2):.4f}")
print(f"P(2 <= X <= 4) = {probability(blood_units, lower=2, upper=4):.4f}")
```

### 2. Grenswaarden uitrekenen

```python
g = boundary_values(blood_units, area=0.05, tail="right")
print(f"Grenswaarde: {g}")
```

### 3. Beschrijvende statistieken berekenen

```python
print(f"$E(X) = {compute_expected_value(blood_units):.4f}$")
print(f"$Var(X) = {compute_variance(blood_units):.4f}$")
print(f"$\\sigma(X) = {compute_standard_deviation(blood_units):.4f}$")
```

## Bekende discrete verdelingen

### Binomiale verdeling

```python
X = get_distribution("binomial", n=10, p=0.25)

print(f"P(X = 3) = {probability(X, lower=3, upper=3):.4f}")
print(f"P(X <= 3) = {probability(X, upper=3):.4f}")
print(f"P(X >= 3) = {probability(X, lower=3):.4f}")

g = boundary_values(X, area=0.05, tail="right")
print(f"Grenswaarde: {g}")

print(f"$E(X) = {X.mean():.4f}$")
print(f"$Var(X) = {X.var():.4f}$")
print(f"$\\sigma(X) = {X.std():.4f}$")
```

### Poissonverdeling

```python
X = get_distribution("poisson", mu=2.5)

print(f"P(X <= 3) = {probability(X, upper=3):.4f}")
print(f"P(X >= 5) = {probability(X, lower=5):.4f}")
```

## Willekeurige continue verdelingen

```python
exponential = CustomContinuousDistribution(
    pdf=lambda x: 0.5 * np.exp(-0.5 * x),
    support=(0, np.inf),
)

# Bereken kansen
p = probability(exponential, lower=2, upper=5)
print(f"P(2 <= X <= 5) = {p:.4f}")

# Bereken grenswaarden
g = boundary_values(exponential, area=0.05, tail="right")
print(f"P(X >= {g:.4f}) = 0.05")
```
