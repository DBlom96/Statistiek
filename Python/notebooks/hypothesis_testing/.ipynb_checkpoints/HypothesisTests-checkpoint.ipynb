```python
import numpy as np
from IPython.display import display, Markdown

# Importeer je eigen packages vanuit statistics_utils
from statistics_utils.calculations.hypothesis_tests import HypothesisTestRunner
from statistics_utils.reporting.hypothesis_tests import report_t_test_one_sample

def simulate_student_exam_question(student_id=42):
    """
    Genereert een unieke tentamenopgave en direct de LaTeX-correctiesleutel
    gebaseerd op het studentnummer als random seed.
    """
    # 1. Deterministische randomisatie per student
    np.random.seed(student_id)
    
    # Gegeven opgaveparameters
    mu_0 = 50.0  # Nulhypothese waarde (bijv. verwachte levensduur in uren)
    alpha = 0.05
    sample_size = 12
    
    # Genereer realistische steekproefdata rondom de ware (onbekende) gemiddelden
    sample_data = np.random.normal(loc=48.5, scale=3.2, size=sample_size)
    
    # 2. Voer de toets uit via jouw statistische engine
    result = HypothesisTestRunner.t_test_one_sample(
        mu_0=mu_0, 
        sample=sample_data, 
        alpha=alpha, 
        test_type='left-sided'
    )
    
    # 3. Genereer het LaTeX-rapportageblok via je reporting module
    latex_snippet = report_t_test_one_sample(result, alpha=alpha)
    
    # 4. Toon de opgave en uitwerking netjes in de Jupyter output
    display(Markdown(f"### Opgave voor Student #{student_id}"))
    display(Markdown(f"""
Een eenheid onderzoekt of de gemiddelde batterijduur van een nieuw type versleutelde radio onder de norm van $\\mu_0 = {mu_0}$ uur ligt (linkszijdige toets). 
De volgende steekproef van $n = {sample_size}$ eenheden is gemeten:
$$ \\mathbf{{x}} = \\{{{', '.join([f'{x:.2f}' for x in sample_data])}\\}} $$
Toets deze hypothese met een significantieniveau van $\\alpha = {alpha}$.
    """))
    
    display(Markdown("---"))
    display(Markdown("### Correctiesleutel & Uitwerking (LaTeX)"))
    print(latex_snippet)

# Test de generator direct in je notebook:
simulate_student_exam_question(student_id=1042)
```
