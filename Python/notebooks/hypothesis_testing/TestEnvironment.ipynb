```python
from scipy.stats import norm, t, binom, poisson, chi2, f
import numpy as np
```

```python
def determine_z_value(opp):
    return norm.ppf(q=opp)

alpha = 0.04
print(determine_z_value(1-alpha/2))
```

```python
sample_mean = 34.6
sample_std = 3.04
alpha = 0.04
n = 25

tvalue = t.ppf(q=1-alpha/2, df=n-1)

left = sample_mean - tvalue * sample_std / np.sqrt(n)
right = sample_mean + tvalue * sample_std / np.sqrt(n)
print(f"The \\SI{{{int(100 * (1 - alpha))}}}-confidence interval for the mean $\\mu$ is $[{left:.4f}, {right:.4f}]$")
```
