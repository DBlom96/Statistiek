# Student's $t$-verdeling

In de volgende functies gaan we ervan uit dat de parameters mu en sigma van de normale verdeling NIET bekend zijn.
Beide parameters moeten worden geschat aan de hand van een steekproefschatting.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, t
```

## 1. Gegeven grenswaarden $a$ en $b$, bereken de kans op uitkomst tussen $a$ en $b$

```python
def compute_probabilities(left=None, right=None, df=1):
    # Kansberekening
    if left is not None and right is not None:
        assert left <= right
        prob = t.cdf(right, df=df) - t.cdf(left, df=df)
        print(f"P({left} <= T <= {right}) = {prob:.4f}")
        return prob
    elif left is not None:
        prob = 1 - t.cdf(left, df=df)
        print(f"P(T >= {left}) = {prob:.4f}")
        return prob
    else:
        prob = t.cdf(right, df=df)
        print(f"P(T <= {right}) = {prob:.4f}")
        return prob

compute_probabilities(left=0, right=1, df=1)
```

## 2. Gegeven kans $p$, bereken de grenswaarden behorende bij deze kans $p$

```python
def compute_boundary_values(area=0.05, df=1, tail="left"):
    if tail == "left":
        g = t.ppf(area, df=df)
        print(f"P(T <= g) = {area} geeft g = {g:.4f}")
        return g
    elif tail == "center":
        g1, g2 = t.ppf(area/2, df=df), t.ppf(1-area/2, df=df)
        print(f"P(g1 <= T <= g2) = {area} geeft g1 = {g1:.4f} en g2 = {g2:.4f}")
        return g1, g2
    else:
        g = t.ppf(1-area, df=df)
        print(f"P(T >= g) = {area} geeft g = {g:.4f}")
        return g

compute_boundary_values(area=0.05, df=9, tail="center")
```

## 3. Gegeven een steekproefgemiddelde $\bar{x}$, en steekproefgrootte $n$, bereken het betrouwbaarheidsinterval

```python
def compute_confidence_t_interval(data=None, sample_mean=None, sample_std=None, n=None, confidence=0.05, tail="center"):
    if data is not None:
        assert isinstance(data, list)
        n = len(data)
        sample_mean = sum(data) / len(data)
        sample_std = np.sqrt(sum( (k - sample_mean) ** 2 for k in data) / (n - 1))
    else:
        assert sample_mean is not None and sample_std is not None and n is not None
    area = 1 - confidence
    std_error = sample_std / np.sqrt(n)
    df = n - 1

    text = f"\\SI{{{int(100 * area)}}}{{\\percent}}-betrouwbaarheidsinterval voor het gemiddelde $\\mu$ ($\\sigma$ onbekend) "
    # Kansberekening
    if tail == "left":
        tval = t.ppf(confidence, df=df)
        g = sample_mean + tval * std_error
        print(text + f"(linkszijdig): $[{g:.4f}, \\infty)$")
        return g
    elif tail == "center":
        tval1 = t.ppf(confidence / 2, df=df)
        tval2 = t.ppf(1 - confidence / 2, df=df)
        g1 = sample_mean + tval1 * std_error
        g2 = sample_mean + tval2 * std_error
        print(text + f"(tweezijdig): $[{g1:.4f}, {g2:.4f}]$")
        return g1, g2
    elif tail == "right":
        tval = t.ppf(1 - confidence, df=df)
        g = sample_mean + tval * std_error
        print(text + f"(rechtszijdig): $(-\\infty, {g:.4f}]$")
        return g

print("Exercise 8.37:")
data = [2216, 2237, 2249, 2204, 2225, 2301, 2281, 2263, 2318, 2255]
compute_confidence_t_interval(data, confidence=0.05)
compute_confidence_t_interval(data, confidence=0.05, tail="left")

print("Exercise 8.39:")
data = [3.775302, 3.350679, 4.217981, 4.030324, 4.639692, 4.139665, 4.395575, 4.824257, 4.268119, 4.584193, 4.930027, 4.315973, 4.600101]
compute_confidence_t_interval(data, confidence=0.05)
compute_confidence_t_interval(data, confidence=0.05, tail="left")

print("Exercise 8.41:")
data = [2.94, 3.00, 2.90, 2.75, 3.00, 2.95, 2.90, 2.75, 2.95, 2.82, 2.81, 3.05]
compute_confidence_t_interval(data, confidence=0.01)
compute_confidence_t_interval(data, confidence=0.01, tail="left")

print("Exercise 9.59:")
data = [97.8, 97.2, 97.4, 97.6, 97.8, 97.9, 98.0, 98.0, 98.0, 98.1, 98.2, 98.3, 98.3, 98.4, 98.4, 98.4, 98.5, 98.6, 98.6, 98.7,
98.8, 98.8, 98.9, 98.9, 99.0]
compute_confidence_t_interval(data, confidence=0.05)
```
