## Descriptieve statistieken

```python
def latex_sample_mean(data):
    n = len(data)
    sample_mean = sum(data) / n
    mean = pretty_print(sample_mean)
    latex_code = f"""
    We berekenen het steekproefgemiddelde $\\overline{{x}}$ als volgt:
    \\begin{{align*}}
        \\overline{{x}} = \\frac{{x_1+x_2+\\ldots+x_n}}{{n}} = \\frac{{ {data[0]} + {data[1]} + \\ldots + {data[-1]} }}{{ {n} }} \\approx {mean}.
    \\end{{align*}}
    """
    return latex_code

def latex_sample_variance(data, ddof=1):
    n = len(data)
    sample_mean = sum(data) / n
    sample_var = np.var(data, ddof=1)
    mean = pretty_print(sample_mean)
    var = pretty_print(sample_var)
    latex_code = f"""
        We berekenen de steekproefvariantie $s^2$ als volgt:
        \\begin{{align*}}
            s^2 &= \\frac{{ (x_1 - \\overline{{x}})^2 + (x_2 - \\overline{{x}})^2 + \\ldots + (x_n - \\overline{{x}})^2 }}{{ n - {ddof} }} \\\\
                &= \\frac{{ ({data[0]} - {mean})^2 + ({data[1]} - {mean})^2 + \\ldots + ({data[-1]} - {mean})^2 }}{{ {n} - {ddof} }} \\\\
                &\\approx {var}.
        \\end{{align*}}
    """
    return latex_code

def latex_sample_std(data, ddof=1):
    sample_var = np.var(data, ddof=1)
    var = pretty_print(sample_var)
    std = pretty_print(np.sqrt(sample_var))
    latex_code = f"""
        We berekenen de steekproefstandaardafwijking $s$ door te wortel van de steekproefvariantie $s^2$ te nemen:
        \\begin{{align*}}
            s = \\sqrt{{ s^2 }} = \\sqrt{{ {var} }} \\approx {std}.
        \\end{{align*}}
    """
    return latex_code

print("\nExercise 9.130")
data = [220, 225, 297, 315, 282, 313]
print(latex_sample_mean(data))
print(latex_sample_variance(data))
print(latex_sample_std(data))
```

## 1. $z$-toets voor het gemiddelde $\mu$ van een normale verdeling (als $\sigma$ bekend is)

### (a) Kritiek gebied berekenen

```python
def compute_critical_region_ztest(mu_0=0, data=None, n=None, sigma=1, confidence=0.05, test_type="tweezijdig"):
    if data is not None:
        assert isinstance(data, list)
        n = len(data)
    else:
        assert n is not None
    area = 1 - confidence
    std_error = sigma / np.sqrt(n)

    # Kansberekening
    if test_type == "linkszijdig":
        critical_value_left = norm.ppf(confidence, mu_0, std_error)
        return critical_value_left, None
    elif test_type == "tweezijdig":
        critical_value_left = norm.ppf(confidence / 2, mu_0, std_error)
        critical_value_right = norm.ppf(1 - confidence / 2, mu_0, std_error)
        return critical_value_left, critical_value_right
    elif test_type == "rechtszijdig":
        critical_value_right = norm.ppf(1 - confidence, mu_0, std_error)
        return None, critical_value_right

def latex_critical_region_ztest(mu_0=0, data=None, n=None, sigma=1, confidence=0.05, test_type="tweezijdig"):
    critical_value_left, critical_value_right = compute_critical_region_ztest(mu_0, data, n, sigma, confidence, test_type)
    if test_type == "linkszijdig":
        vorm = "$Z = (-\\infty, C_L]$"
    elif test_type == "tweezijdig":
        vorm = "$Z = (-\\infty, C_L] \\cup [C_R, \\infty)$"
    elif test_type == "rechtszijdig":
        vorm = "$Z = [C_R, \\infty)$"
    print(f"Het kritieke gebied ($\\alpha = {confidence}$, {test_type}e $z$-toets) is van de vorm {vorm}, waarbij:")
    print("\\begin{align*}")
    if test_type == "linkszijdig":
        print(f"C_L &= \\invnorm(\\opp=\\alpha={confidence}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {critical_value_left:.4f}")
    elif test_type == "tweezijdig":
        print(f"\tC_L &= \\invnorm(\\opp=\\alpha/2 = {confidence/2}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {critical_value_left:.4f} \\\\")
        print(f"\tC_R &= \\invnorm(\\opp=1-\\alpha/2 = {1-confidence/2}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {critical_value_right:.4f}")
    elif test_type == "rechtszijdig":
        print(f"\tC_R &= \\invnorm(\\opp=1-\\alpha = {1-confidence}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {critical_value_right:.4f}")
    print("\\end{align*}")
    

    
print("Exercise 9.42")
latex_critical_region_ztest(mu_0=68.3, data=None, n=15, sigma=0.83, confidence=0.01, test_type="tweezijdig")

print("\nExercise 9.47")
latex_critical_region_ztest(mu_0=4, data=None, n=50, sigma=0.2, confidence=0.05, test_type="rechtszijdig")

print("\nExercise 9.131")
latex_critical_region_ztest(mu_0=85, data=None, n=30, sigma=16, confidence=0.01, test_type="rechtszijdig")
latex_critical_region_ztest(mu_0=85, data=None, n=100, sigma=16, confidence=0.01, test_type="rechtszijdig")
latex_critical_region_ztest(mu_0=85, data=None, n=400, sigma=16, confidence=0.01, test_type="rechtszijdig")
latex_critical_region_ztest(mu_0=85, data=None, n=2550, sigma=16, confidence=0.01, test_type="rechtszijdig")
```

### (b) $p$-waardes berekenen

```python
def compute_p_value_ztest(mu_0=0, data=None, sample_mean=None, n=None, sigma=1, test_type="tweezijdig"):
    if data is not None:
        assert isinstance(data, list)
        n = len(data)
        sample_mean = sum(data) / len(data)
    else:
        assert sample_mean is not None and n is not None
    std_error = sigma / np.sqrt(n)
    text = f"De $p$-waarde behorende bij $\\overline{{x}} = {sample_mean:.4f}$ is $p = "

    # Kansberekening
    if test_type == "linkszijdig":
        p_waarde = norm.cdf(sample_mean, mu_0, std_error)
        print(text + f"P(\\overline{{X}} \\leq {sample_mean:.4f}) = \\normalcdf(\\lb=-\\GRinfinity, \\ub={sample_mean:.4f}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {p_waarde:.4f}$")
    elif test_type == "tweezijdig":
        if sample_mean >= mu_0:
            p_waarde = 2 * (1 - norm.cdf(sample_mean, mu_0, std_error))
            print(text + f"2 \\cdot P(\\overline{{X}} \\geq {sample_mean:.4f}) = 2 \\cdot \\normalcdf(\\lb={sample_mean:.4f}, \\ub=\\GRinfinity, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {p_waarde:.4f}$")
        else:
            p_waarde = 2 * norm.cdf(sample_mean, mu_0, std_error)
            print(text + f"2 \\cdot P(\\overline{{X}} \\leq {sample_mean:.4f}) = 2 \\cdot \\normalcdf(\\lb=-\\GRinfinity, \\ub={sample_mean:.4f}, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {p_waarde:.4f}$")
    elif test_type == "rechtszijdig":
        p_waarde = 1 - norm.cdf(sample_mean, mu_0, std_error)
        print(text + f"P(\\overline{{X}} \\geq {sample_mean:.4f}) = \\normalcdf(\\lb={sample_mean:.4f}, \\ub=\\GRinfinity, \\mu={mu_0}, \\sigma={sigma}/\\sqrt{{{n}}}) \\approx {p_waarde:.4f}$")

print("Exercise 9.42")
compute_p_value_ztest(mu_0=68.3, data=None, sample_mean=68, n=15, sigma=0.83, test_type="tweezijdig")

print("Exercise 9.47")
compute_p_value_ztest(mu_0=4, data=None, sample_mean=4.05, n=50, sigma=0.2, test_type="rechtszijdig")

print("\nExercise 9.131")
compute_p_value_ztest(mu_0=85, data=None, sample_mean=86, n=30, sigma=16, test_type="rechtszijdig")
compute_p_value_ztest(mu_0=85, data=None, sample_mean=86, n=100, sigma=16, test_type="rechtszijdig")
compute_p_value_ztest(mu_0=85, data=None, sample_mean=86, n=400, sigma=16, test_type="rechtszijdig")
compute_p_value_ztest(mu_0=85, data=None, sample_mean=86, n=2550, sigma=16, test_type="rechtszijdig")
```

### (c) Onderscheidend vermogen ($1 - \beta$) berekenen van een hypothesetoets

```python
def compute_power_ztest(mu_0=0, mu_1=1, data=None, n=None, sigma=1, confidence=0.05, test_type="tweezijdig"):
    critical_value_left, critical_value_right = compute_critical_region_ztest(mu_0, data, n, sigma, confidence, test_type)
    if data is not None:
        assert isinstance(data, list)
        n = len(data)
    else:
        assert n is not None
    area = 1 - confidence
    std_error = sigma / np.sqrt(n)

    text = f"Het onderscheidend vermogen van deze hypothesetoets ($\\alpha = {confidence}$, {test_type}e toets) is gelijk aan "
    print("\\begin{{align*}}")
    if test_type == "linkszijdig":
        power = norm.cdf(critical_value_left, mu_1, std_error)
        print(f"1 - \\beta = P(H_1 \\mid H_1) &= P(\\overline{{X}} \\leq C_L \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= P(\\overline{{X}} \\leq {critical_value_left:.4f} \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= \\normalcdf(\\lb=-\\GRinfinity, \\ub={critical_value_left:.4f}, \\mu = {mu_1}, \\sigma = {sigma}/\\sqrt{{{n}}}) \\\\")
        print(f"                              &\\approx {power:.4f}")
    elif test_type == "tweezijdig":
        power = norm.cdf(critical_value_left, mu_1, std_error) + (1 - norm.cdf(critical_value_right, mu_1, std_error))
        print(f"1 - \\beta = P(H_1 \\mid H_1) &= P(\\overline{{X}} \\leq C_L \\mid \\mu = {mu_1}) + P(\\overline{{X}} \\geq C_R \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= P(\\overline{{X}} \\leq {critical_value_left:.4f} \\mid \\mu = {mu_1}) + P(\\overline{{X}} \\geq {critical_value_right:.4f} \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= \\normalcdf(\\lb=-\\GRinfinity, \\ub={critical_value_left:.4f}, \\mu = {mu_1}, \\sigma = {sigma}/\\sqrt{{{n}}}) + \\normalcdf(\\lb={critical_value_right:.4f}, \\ub=\\GRinfinity, \\mu = {mu_1}, \\sigma = {sigma}/\\sqrt{{{n}}}) \\\\")
        print(f"                              &\\approx {power:.4f}")
    elif test_type == "rechtszijdig":
        power = 1 - norm.cdf(critical_value_right, mu_1, std_error)
        print(f"1 - \\beta = P(H_1 \\mid H_1) &= P(\\overline{{X}} \\geq C_R \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= P(\\overline{{X}} \\geq {critical_value_right:.4f} \\mid \\mu = {mu_1}) \\\\")
        print(f"                              &= \\normalcdf(\\lb={critical_value_right:.4f}, \\ub=\\GRinfinity, \\mu = {mu_1}, \\sigma = {sigma}/\\sqrt{{{n}}}) \\\\")
        print(f"                              &\\approx {power:.4f}")
    print("\\end{{align*}}")

print("Exercise 9.42")
compute_power_ztest(mu_0=68.3, mu_1=65.5, data=None, n=15, sigma=0.83, confidence=0.01, test_type="tweezijdig")

print("\nExercise 9.47")
compute_power_ztest(mu_0=4, mu_1=4.5, data=None, n=50, sigma=0.2, confidence=0.05, test_type="rechtszijdig")

print("\nExercise 9.131")
compute_power_ztest(mu_0=85, mu_1=86, data=None, n=30, sigma=16, confidence=0.01, test_type="rechtszijdig")
compute_power_ztest(mu_0=85, mu_1=86, data=None, n=100, sigma=16, confidence=0.01, test_type="rechtszijdig")
compute_power_ztest(mu_0=85, mu_1=86, data=None, n=400, sigma=16, confidence=0.01, test_type="rechtszijdig")
compute_power_ztest(mu_0=85, mu_1=86, data=None, n=2550, sigma=16, confidence=0.01, test_type="rechtszijdig")
```

## 3. Chikwadraat goodness-of-fit test

### (a) Kritiek gebied berekenen

```python
def compute_critical_region_chisq(data=None, num_values=None, num_estimated_params=0, confidence=0.05):
    if data is not None:
        assert isinstance(data, list)
        num_values = len(data)
    else:
        assert num_values is not None

    # Chikwadraat goodness-of-fit test is altijd rechtszijdig
    critical_value_right = chi2.ppf(q=1 - confidence, df=num_values-num_estimated_params-1)
    return None, critical_value_right

def latex_critical_region_chisqtest(data=None, num_values=None, num_estimated_params=0, confidence=0.05):
    critical_value_left, critical_value_right = compute_critical_region_chisq(data, num_values, num_estimated_params, confidence)
    num_values = num_values if num_values is not None else len(data)
    print(f"""Het kritieke gebied ($\\alpha = {confidence}$, $\\chi^2$-toets) is van de vorm $Z = [C_R, \\infty)$, waarbij we de
    waarde van $C_R$ kunnen berekenen met behulp van de \\enquote{{numerieke solver}}:""")
    print("\\begin{align*}")
    print(f"\ty_1 &= \\chi^2\\text{{cdf}}(\\lb=X, \\ub=\\GRinfinity, \\df={num_values}-{num_estimated_params}-1 = {num_values-num_estimated_params-1}) \\\\")
    print(f"\ty_2 &= \\alpha = {confidence}")
    print("\\end{align*}")
    print(f"De numerieke solver geeft een waarde $C_R \\approx {critical_value_right:.4f}$")
    
print("Exercise 9.96")
data = [24,30,31,11,4]
latex_critical_region_chisqtest(data=data, num_values=None, num_estimated_params=0, confidence=0.05)

print("\nExercise 9.97")
data = [12,8,13,11,12,10,9]
latex_critical_region_chisqtest(data=data, num_values=None, num_estimated_params=1, confidence=0.05)

print("Exercise 9.99")
data = [4,14,10,22]
latex_critical_region_chisqtest(data=data, num_values=None, num_estimated_params=0, confidence=0.05)
```
